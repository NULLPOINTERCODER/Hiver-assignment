# 04. Baseline Models & Final Classifier Benchmark
Compares Majority Class, TF-IDF + Logistic, and Semantic Classifier on 200 Golden samples.


```python
import json
import pandas as pd

with open('../results/baseline_results.json', 'r') as f:
    results = json.load(f)

summary = []
for model_name, m in results.items():
    summary.append({
        'Model': model_name,
        'Accuracy': m['accuracy'],
        'Macro Precision': m['macro_precision'],
        'Macro Recall': m['macro_recall'],
        'Macro F1': m['macro_f1']
    })

pd.DataFrame(summary)
```
