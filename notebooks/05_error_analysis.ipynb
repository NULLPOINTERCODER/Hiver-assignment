# 05. Error Analysis & Failure Modes
Deep diagnostic analysis of prediction errors, false auto-handling cases, and confidence distributions.


```python
import json
import pandas as pd

preds = []
with open('../evaluation/predictions.jsonl', 'r') as f:
    for line in f:
        preds.append(json.loads(line))

pred_df = pd.DataFrame(preds)
print('Intent Match Rate:', pred_df['intent_match'].mean())
print('Action Match Rate:', pred_df['action_match'].mean())

# Top misclassifications
errors = pred_df[~pred_df['intent_match']]
print(f'Total errors: {len(errors)}')
errors[['id', 'expected_intent', 'predicted_intent', 'customer_message']].head(5)
```
