# 06. Human vs LLM Judge Agreement Analysis
Measures inter-rater agreement (Spearman rho, Pearson r, Cohen's kappa) on 50 audited samples.


```python
import json
import pandas as pd

with open('../evaluation/metrics.json', 'r') as f:
    metrics = json.load(f)

agreement = metrics.get('human_judge_agreement', {})
print('Human-Judge Correlation Across Dimensions:')
for dim, res in agreement.items():
    print(f"- {dim}: Spearman rho = {res.get('spearman_correlation')}, Exact Match = {res.get('exact_agreement_rate')*100:.1f}%")
```
